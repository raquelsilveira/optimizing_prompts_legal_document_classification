```python
import dsp
import dspy
from dspy.signatures.signature import ensure_signature

from dspy.predict import Predict

class ChainOfThought(Predict):
    def __init__(self, signature, description = '', rationale_type=None, activated=True, **config):
        super().__init__(signature, **config)

        self.activated = activated

        signature = ensure_signature(self.signature)
        *_keys, last_key = signature.output_fields.keys()

        rationale_type = rationale_type or dspy.OutputField(
            prefix="Raciocínio: Vamos pensar passo a passo em ordem para",
            desc=description,
        )

        self.extended_signature = signature.prepend("rationale", rationale_type, type_=str)

    def forward(self, **kwargs):
        new_signature = kwargs.pop("new_signature", None)
        if new_signature is None:
            if self.activated is True or (
                self.activated is None and isinstance(dsp.settings.lm, dsp.GPT3)
            ):
                signature = self.extended_signature
            else:
                signature = self.signature
        else:
            signature = new_signature
        return super().forward(signature=signature, **kwargs)


    def dump_state(self):
        state = super().dump_state()

        # Cache the signature instructions and the last field's name.
        state["extended_signature_instructions"] = self.extended_signature.instructions

        *_, last_key = self.signature.fields.keys()
        state["extended_signature_prefix"] = self.extended_signature.fields[last_key].json_schema_extra['prefix']

        return state

    def load_state(self, state):
        super().load_state(state)

        # Reconstruct the signature.
        if "extended_signature_instructions" in state:
            instructions = state["extended_signature_instructions"]
            self.extended_signature = self.extended_signature.with_instructions(instructions)

        if "extended_signature_prefix" in state:
            prefix = state["extended_signature_prefix"]
            *_, last_key = self.extended_signature.fields.keys()
            self.extended_signature = self.extended_signature.with_updated_fields(last_key, prefix=prefix)
```

## População Inicial

```python
class InitialPopulationSignature(Signature):

    instruction = dspy.InputField(desc="Instrução Inicial:")
    thinking_style = dspy.InputField(desc="Thinking Style:")
    mutation_prompt = dspy.InputField(desc="Mutation Prompt:")

    mutated_prompt = dspy.OutputField(desc="Instrução Mutante:")

class InitialPopulationModule(dspy.Module):
    def __init__(self):
        super().__init__()
        self.initial_population = ChainOfThought(InitialPopulationSignature, description='${produzir a Instrução Mutante}, tomando como base a Instrução Inicial, o Thinking Style e o Mutation Prompt...')

    def forward(self, mutation_prompt, thinking_style, initial_instruction):

        prediction = self.initial_population(instruction=initial_instruction, thinking_style=thinking_style, mutation_prompt=mutation_prompt)
        prediction['instruction'] = initial_instruction
        prediction['thinking_style'] = thinking_style
        prediction['mutation_prompt'] = mutation_prompt

        return prediction

initial_population = InitialPopulationModule()
```

## Estimation of Distribution Mutation

```python
import dspy

class EDAMutationSignature(Signature):

    prompt_list = dspy.InputField(desc="Lista de instruções:")
    mutated_prompt = dspy.OutputField(desc="Instrução Mutante:")

class EDAMutationModule(dspy.Module):
    def __init__(self):
        super().__init__()
        self.eda_mutation = ChainOfThought(EDAMutationSignature, description='${produzir uma Instrução Mutante melhor}, tomando como base uma lista de instruções...')

    def forward(self, prompt_list):

        eda_prompt = "Lista de instruções:\n" + "\n".join(prompt_list)
        prediction = self.eda_mutation(prompt_list=eda_prompt)
        prediction['mutated_prompt'] = prediction['mutated_prompt'].replace('Instrução Mutante:', '').strip()
        return prediction

eda_mutation_operator = EDAMutationModule()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer

SIMILARITY_THRESHOLD = 0.95

def get_eda_prompts(prompts_tasks, population_fitness):
    model = SentenceTransformer('raquelsilveira/legalbertpt_fp')
    embeddings = model.encode(prompts_tasks)
    embeddings_similarity = cosine_similarity(embeddings)
    
    filtered_indices = []
    for i in range(len(embeddings)):
        if not any(embeddings_similarity[i, :i] > SIMILARITY_THRESHOLD):
            filtered_indices.append(i)

    filtered_prompt_units = [(prompts_tasks[i], population_fitness[i]) for i in filtered_indices]
    return filtered_prompt_units
```

```python
class HyperMutationFirstOrderSignature(Signature):
    hypermutation_prompt = dspy.InputField(desc="HyperMutation Prompt:")
    mutation_prompt = dspy.InputField(desc="Mutation Prompt:")

    new_mutation_prompt = dspy.OutputField(desc="Novo Mutation Prompt:")

class HyperMutationFirstOrderModule(dspy.Module):
    def __init__(self):
        super().__init__()
        self.hyper_mutation = ChainOfThought(HyperMutationFirstOrderSignature, description='${produzir um Novo Mutation Prompt}, tomando como base o Hypermutation e o Mutation Prompts ...')

    def forward(self, mutation_prompt):
        hypermutation_prompt = 'Por favor, resuma e melhore a seguinte instrução:'
        prediction = self.hyper_mutation(hypermutation_prompt=hypermutation_prompt, mutation_prompt=mutation_prompt)
        
        prediction['new_mutation_prompt'] = prediction['new_mutation_prompt'].replace('Novo Mutation Prompt:', '').strip()
      
        return prediction

hyper_mutation_first_order = HyperMutationFirstOrderModule()
```

```python
class LamarckianMutationSignature(Signature):
    lamarckian_prompt = dspy.InputField(desc="Lamarckian prompt:")
    mutated_prompt = dspy.OutputField(desc="Instrução Mutante:")

class LamarckianMutationModule(dspy.Module):
    def __init__(self):
        super().__init__()
        self.lamarckian_mutation = ChainOfThought(LamarckianMutationSignature, description='${produzir a Instrução Mutante}, tomando como base o Lamarckian prompt...')

    def forward(self, exemplo_positivo, exemplo_negativo):

      lamarckian_prompt = f'''Dei uma instrução a um amigo para classificar com precisão se um documento jurídico é um {tipo_documento_resumo} ou não. 
                          O amigo leu a instrução e escreveu a saída de cada entrada. Aqui estão os pares de entrada e saída: '''
      
      lamarckian_prompt += f'\nEntrada: {exemplo_positivo}.\nSaída: {tipo_documento_resumo}.\n'
      lamarckian_prompt += f'\nEntrada: {exemplo_negativo}.\nSaída: Não é {tipo_documento_resumo}.\n'
      
      lamarckian_prompt += '\nA instrução foi:'

      prediction = self.lamarckian_mutation(lamarckian_prompt=lamarckian_prompt)
      
      prediction['mutated_prompt'] = prediction['mutated_prompt'].replace('Instrução Mutante:', '').strip()
      
      return prediction

lamarckian_mutation_operator = LamarckianMutationModule()
```

```python
class PromptCrossoverSignature(Signature):
    
    '''Junte os elementos das Instruções 1 e 2 para gerar uma instrução melhor, que tem como ÚNICA TAREFA identificar se um texto jurídico é do
    tipo despacho saneador ou não.'''
    
    instrucao_1 = dspy.InputField(desc="Instrução 1:")
    instrucao_2 = dspy.InputField(desc="Instrução 2:")

    mutated_prompt = dspy.OutputField(desc="Instrução Mutante:")

class PromptCrossoverModule(dspy.Module):
    def __init__(self):
        super().__init__()
        self.crossover_mutation = ChainOfThought(PromptCrossoverSignature, description='${produzir uma Instrução Mutante melhor}, tomando como base duas intruções da tarefa fornecidas...')

    def forward(self, instrucao_1, instrucao_2):
        prediction = self.crossover_mutation(instrucao_1=instrucao_1, instrucao_2=instrucao_2)
        prediction['mutated_prompt'] = prediction['mutated_prompt'].replace('Instrução Mutante:', '').strip()
        return prediction

prompt_crossover_module = PromptCrossoverModule()
```

```python
import pydantic
from typing import List

class EvolutionUnit(pydantic.BaseModel):
    """ A individual unit of the overall population.

    Attributes:
        'thinking_style': the thinking_style prompt.

        'mutation_prompt': the mutation_prompt.

        'task_prompt': the task_prompt.

        'fitness': the estimated performance of the unit.

        'history': historical prompts for analysis.
    """
    id: str
    parent: str
    thinking_style: str
    mutation_prompt: str
    task_prompt: str
    fitness: float
    f1_macro: float
    f1_positive: float
    history: List[str]
    mutator_operator: str
    crossover_operator: str

class Population(pydantic.BaseModel):
    """ Population model that holds the age of the population, its size, and a list of individuals.

    Attributes:
        'size' (int): the size of the population.

        'age' (int): the age of the population.

        'units' (List[EvolutionUnit]): the individuals of a population.
    """
    size: int
    age: int
    problem_description: str
    elites: List[EvolutionUnit]
    units: List[EvolutionUnit]
```

```python
def create_population(size: int, problem_description: str) -> Population:
    """samples the mutation_prompts and thinking_styles and returns a 'Population' object.

    Args:
        'size' (int): the size of the population to create.
        'problem_description (D)' (str): the problem description we are optimizing for.
    """
    data = {
        'size': size,
        'age': 0,
        'problem_description' : problem_description,
        'elites' : [],
        'units': [EvolutionUnit(**{
            'id': '',
            'parent': '',
            'thinking_style' : random.choice(THINKING_STYLES_PROMPTS),
            'mutation_prompt' : random.choice(MUTATION_PROMPTS),
            'task_prompt' : '',
            'fitness' : -1,
            'f1_macro': -1,
            'f1_positive': -1,
            'history' : [],
            'mutator_operator': '',
            'crossover_operator': ''
            }) for i in range(size)]
    }

    return Population(**data)
```

```python
import random
from random import choices
import re
import os
from typing import List
from sentence_transformers import SentenceTransformer, util
from itertools import chain
from math import exp

random.seed(42)
model = SentenceTransformer('raquelsilveira/legalbertpt_fp')
similarity_threshold = 0.95


def zero_order_prompt_gen(unit: EvolutionUnit, unit_replace: EvolutionUnit, **kwargs) -> EvolutionUnit:

  generate_inst_mut = direct_mutation_zero_order(problem=instrucoes_iniciais[tipo_documento])
  pattern = r"1\.(.*?)2\."
  match = re.search(pattern, generate_inst_mut.dicas, re.DOTALL)
  if match:
    unit_replace.task_prompt = f'{instrucoes_iniciais[tipo_documento]} {match.group(1).strip()}'
  else:
    unit_replace.task_prompt = instrucoes_iniciais[tipo_documento]

  return unit_replace


def first_order_prompt_gen(unit: EvolutionUnit, unit_replace: EvolutionUnit, **kwargs) -> EvolutionUnit:

    unit_replace.task_prompt = direct_mutation(instruction=unit.task_prompt,
                                                mutation_prompt=unit.mutation_prompt
                                              ).task_prompt
    return unit_replace
  

def eda_mutation(unit: EvolutionUnit, unit_replace: EvolutionUnit, **kwargs) -> EvolutionUnit:

  population = kwargs.get('population').units
  prompts_tasks, population_fitness = [], []
  for p in population:
    prompts_tasks.append(p.task_prompt)
    population_fitness.append(p.fitness)
  
  filtered_prompt_units = get_eda_prompts(prompts_tasks, population_fitness)

  random.shuffle(filtered_prompt_units)

  shuffled_task_prompts = "\n".join(
    [unit[0] for unit in filtered_prompt_units]
  )

  unit_replace.task_prompt = eda_mutation_operator(prompt_list=shuffled_task_prompts).mutated_prompt

  return unit_replace


def eda_rank_mutation(unit: EvolutionUnit, unit_replace: EvolutionUnit, **kwargs) -> EvolutionUnit:

  population = kwargs.get('population').units
  prompts_tasks, population_fitness = [], []
  for p in population:
    prompts_tasks.append(p.task_prompt)
    population_fitness.append(p.fitness)
  
  filtered_prompt_units = get_eda_prompts(prompts_tasks, population_fitness)
  filtered_prompt_units.sort(key=lambda x: x[1])
  sorted_prompt_tasks = [unit[0] for unit in filtered_prompt_units]
  unit_replace.task_prompt = eda_rank_mutation_operator(prompt_list=sorted_prompt_tasks).mutated_prompt

  return unit_replace


def lineage_mutation(unit: EvolutionUnit, unit_replace: EvolutionUnit, **kwargs) -> EvolutionUnit:

  elites = kwargs.get('population').elites
  unit_replace.task_prompt = lineage_mutation_operator(instructions_elites=elites).mutated_prompt

  return unit_replace


def zero_order_hyper_mutation(unit: EvolutionUnit, unit_replace: EvolutionUnit, **kwargs) -> EvolutionUnit:

  hyper_mutation = hyper_mutation_zero_order(instrucoes_iniciais[tipo_documento])
  
  unit_replace.mutation_prompt = hyper_mutation.new_mutation_prompt
  unit_replace.task_prompt = direct_mutation(instruction=unit.task_prompt,
                                            mutation_prompt=hyper_mutation.new_mutation_prompt
                                            ).task_prompt
  return unit_replace


def first_order_hyper_mutation(unit: EvolutionUnit, unit_replace: EvolutionUnit, **kwargs) -> EvolutionUnit:

    hyper_mutation = hyper_mutation_first_order(unit.mutation_prompt)
    unit_replace.mutation_prompt = hyper_mutation.new_mutation_prompt
    unit_replace.task_prompt = direct_mutation(instruction=unit.task_prompt,
                                              mutation_prompt=hyper_mutation.new_mutation_prompt
                                              ).task_prompt
    return unit_replace


def lamarckian_mutation(unit: EvolutionUnit, unit_replace: EvolutionUnit, **kwargs) -> EvolutionUnit:

  exemplo_positivo = df_dataset[df_dataset.label_revisada == tipo_documento].sample(n=1).iloc[0].texto
  exemplo_negativo = df_dataset[df_dataset.label_revisada != tipo_documento].sample(n=1).iloc[0].texto

  unit_replace.task_prompt = lamarckian_mutation_operator(exemplo_positivo, 
                                                          exemplo_negativo).mutated_prompt
  return unit_replace


def softmax(x: List[float]) -> List[float]:
    denom = sum([exp(float(y)) for y in x])
    return [exp(float(y)) / denom for y in x]
  

def prompt_crossover(unit: EvolutionUnit, unit_replace: EvolutionUnit, **kwargs) -> EvolutionUnit:

  units = kwargs.get('population').units
  
  members = [x.task_prompt for x in units if x != unit]
  fitnesses = [x.fitness for x in units if x != unit]
  probabilities = softmax([float(f) for f in fitnesses])

  replacement = choices(members, probabilities)[0]
  unit_replace.task_prompt = prompt_crossover_module(instrucao_1=replacement, instrucao_2 = unit.task_prompt).mutated_prompt
  return unit_replace
  

# mutator operators
MUTATORS = [
    zero_order_prompt_gen,
    first_order_prompt_gen,
    eda_mutation,
    eda_rank_mutation,
    lineage_mutation,
    zero_order_hyper_mutation,
    first_order_hyper_mutation,
    lamarckian_mutation
]

POST_MUTATORS = [
     prompt_crossover,
]

def mutate(population: Population) -> Population:
    """Select and apply a random mutator"""
    # steps
    # 1. parse through the population, grouping each evo unit by 2
    # 2. for each pair of evo units, using a uniform distribution, select a random mutator (of the 9)
    # 3. mutate and populate population.units

    # make index pairs
    indices = [i for i in range(len(population.units))]
    random.shuffle(indices)
    pairs = [indices[2*x:2*x+2] for x in range(len(indices) // 2)]

    # binary tourmanent genetic algorithm
    for i in range(len(pairs)):

        first_unit = population.units[pairs[i][0]]
        second_unit = population.units[pairs[i][1]]

        # determine which unit has the higher fitness. Since I am currently testing and want to preserve the # of calls I am making to the LLM, there
        # is a decent chance that I will hit equal fitness levels. in that case, first unit wins and second unit loses.

        if first_unit.fitness >= second_unit.fitness:
            replace_input = second_unit
            mutation_input = first_unit
        else:
            replace_input = first_unit
            mutation_input = second_unit

        operator_mutation_parent = mutation_input.parent
        replace_input.parent = mutation_input.id
        mutation_input.parent = ''
        mutation_input.mutator_operator = ''

        data = {
            'unit' : mutation_input,
            'unit_replace': replace_input,
            'elites' : population.elites,
            'problem_description': population.problem_description,
            'population': population
        }

        # uniformly pick and call a random mutation operator on the losing unit
        random_mutator = random.sample(MUTATORS, 1)[0]
        
        while random_mutator.__name__ == operator_mutation_parent:
          random_mutator = random.sample(MUTATORS, 1)[0]

        replace_input.mutator_operator = random_mutator.__name__
        replace_input.fitness = -1
        
        random_mutator(**data)
        
        mutation_input.crossover_operator = ''
        replace_input.crossover_operator = ''
        
        random_crossover = random.sample(POST_MUTATORS, 1)[0]
        if random.randint(1, 100) <= 10:
          replace_input.crossover_operator = random_crossover.__name__
          replace_input.fitness = -1
          
          random_crossover(**data)
          
    return population
```

```python
from sklearn import metrics
import time
import tqdm
import matplotlib.pyplot as plt

def init_run(population: Population, instrucao_inicial):
    """ The first run of the population that consumes the prompt_description and
    creates the first prompt_tasks.

    Args:
        population (Population): A population created by `create_population`.
    """

    start_time = time.time()

    results = []
    for id, unit in enumerate(population.units):
        unit.id = f'G0.{id}'
        unit.task_prompt = instrucao_inicial
        r = initial_population(unit.mutation_prompt, population.problem_description, unit.task_prompt)
        results.append(r)

    end_time = time.time()

    print(f"Prompt initialization done. {end_time - start_time}s")

    assert len(results) == population.size, "size of google response to population is mismatched"
    for i, item in enumerate(results):
        population.units[i].task_prompt = item.mutated_prompt

    _evaluate_fitness(0, population)

    return population

def run_for_n(n: int, population: Population):
    """ Runs the genetic algorithm for n generations.
    """
    p = population
    indice_maior, maior_fitness = 0, 0
    
    for i in range(1, n+1):
        print(f"================== Population {i} ================== ")

        for id, u in enumerate(p.units):
            u.id = f'G{i}.{id}'

        mutate(p)
        _evaluate_fitness(i, p)
        
        maior_fitness_populacao = 0
        for u in p.units:
          if u.fitness > maior_fitness_populacao:
            maior_fitness_populacao = u.fitness
            
        if maior_fitness_populacao > maior_fitness:
          maior_fitness = maior_fitness_populacao
          indice_maior = i
          
        #criterio de paciencia
        if i - indice_maior == CRITERIO_PACIENCIA:
          break

    return p
  

def run_for_n_continue(n: int):
    """ Runs the genetic algorithm for n generations.
    """
    # p = population
    indice_maior, maior_fitness = 0, 0
    
    maior_geracao = max(d['generation'] for d in df_data_generation)
    p_atual = [d for d in df_data_generation if d['generation'] == maior_geracao]
    
    data = {
        'size': size,
        'age': 0,
        'problem_description': descricao_problema,
        'elites': [],
        'units': [
            EvolutionUnit(
                id=d.get('id', ''),
                parent=d.get('parent', ''),
                thinking_style=d.get('thinking_style', random.choice(THINKING_STYLES_PROMPTS)),
                mutation_prompt=d.get('mutation_prompt', random.choice(MUTATION_PROMPTS)),
                task_prompt=d.get('task_prompt', ''),
                fitness=d.get('fitness', -1),
                history=d.get('history', []),
                f1_macro=d.get('f1_macro', -1),
                f1_positive=d.get('f1_positive', -1),
                mutator_operator=d.get('mutator_operator', ''),
                crossover_operator=d.get('crossover_operator', '')
            ) for d in p_atual
        ]
    }
    
    p = Population(**data)
    
    for d in df_data_generation:
      if d['fitness'] > maior_fitness:
        maior_fitness = d['fitness']
        indice_maior = d['generation']
    
    for i in range(maior_geracao+1, n+1):

        print(f"================== Population {i} ================== ")

        for id, u in enumerate(p.units):
            u.id = f'G{i}.{id}'

        mutate(p)
        _evaluate_fitness(i, p)
        
        maior_fitness_populacao = 0
        for u in p.units:
          if u.fitness > maior_fitness_populacao:
            maior_fitness_populacao = u.fitness
            
        if maior_fitness_populacao > maior_fitness:
          maior_fitness = maior_fitness_populacao
          indice_maior = i
          
        if i - indice_maior == CRITERIO_PACIENCIA:
          break
    return p
  
def _evaluate_unit(generation, unit):

  print('Evaluate unit...')
  print(unit.task_prompt)
  
  true_label = []
  pred_label = []
  for id, row in tqdm.tqdm(df_data_treino.iterrows()):
      generate_result = classificacao(
                          instruction_task=unit.task_prompt,
                          text=row.texto,
                        )
      pred_label.append(generate_result.avaliacao)
      true_label.append("Sim" if row.label_revisada == tipo_documento else "Não")
      
  unit.fitness = metrics.accuracy_score(true_label, pred_label)
  unit.f1_macro = metrics.f1_score(true_label, pred_label, average='macro')
  unit.f1_positive = metrics.f1_score(true_label, pred_label, pos_label='Sim')
  
  print('='*20, 'Geração: ', generation, '='*20)
  #mostra_matriz_confusao(true_label, pred_label)


def mostra_matriz_confusao(true_label, pred_label):
  cp = metrics.classification_report(true_label, pred_label, output_dict=True)
  labels = list(cp.keys())[:-3]

  print(metrics.classification_report(true_label, pred_label))
  cm = metrics.confusion_matrix(true_label, pred_label)
  metrics.ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=labels).plot()

  plt.show()
  
  
def _evaluate_fitness(generation, population: Population) -> Population:
    """ Evaluates each task-prompt and populates the fitness values.
    """

    print(f"Starting fitness evaluation...")
    start_time = time.time()

    elite_fitness = -1
    for unit_index, unit in enumerate(population.units):
      if unit.fitness == -1:
        _evaluate_unit(generation, unit)
      if unit.fitness > elite_fitness:
        current_elite = unit.model_copy()
        elite_fitness = unit.fitness

    elite_na_populacao = False
    for e in population.elites:
      if e.task_prompt == current_elite.task_prompt:
        elite_na_populacao = True

    if not elite_na_populacao:
      population.elites.append(current_elite)
    end_time = time.time()

    print()
    print('*** Population ***')
    for id, uni in enumerate(population.units):
      print(id)
      print('Thinking style:', uni.thinking_style)
      print('Mutation prompt:', uni.mutation_prompt)
      print('Task prompt:', uni.task_prompt)
      print('Fitness:', uni.fitness)
      print()

    print()
    print('*** Best solution ***')
    prompt = current_elite.task_prompt if current_elite.task_prompt != '' else population.problem_description
    print(f'Prompt: {prompt}')
    print(f'Fitness: {current_elite.fitness}')

    df_data_generation.extend([{'generation': generation,
                                'id': u.id,
                                'thinking_style' : u.thinking_style,
                                'mutation_prompt' : u.mutation_prompt,
                                'task_prompt' : u.task_prompt if u.task_prompt != '' else population.problem_description,
                                'fitness' : u.fitness,
                                'f1_macro': u.f1_macro,
                                'f1_positive': u.f1_positive,
                                'mutator_operator': u.mutator_operator,
                                'parent': u.parent,
                                'crossover_operator': u.crossover_operator
                                }
                               for u in population.units])
    
    print(f"Done fitness evaluation. {end_time - start_time}s")

    return population
```

```python
from sklearn import metrics
import matplotlib.pyplot as plt

def show_evaluation(true_label, pred_label):
    cp = metrics.classification_report(true_label, pred_label, output_dict=True)
    labels = list(cp.keys())[:-3]

    print(metrics.classification_report(true_label, pred_label))
    cm = metrics.confusion_matrix(true_label, pred_label)
    metrics.ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=labels).plot()
    plt.xticks(rotation=90)
    plt.show()

def evaluation_prompt_dspy(task_prompt):
    true_label = []
    pred_label = []
    result = []
    for id, row in tqdm.tqdm(df_data_teste.iterrows()):
        generate_result = classificacao(
                            instruction_task=task_prompt,
                            text=row.texto,
                        )
        pred = generate_result.avaliacao
        true = "Sim" if row.label == tipo_documento else "Não"
        result.append({'id': id,
                'texto': row.texto,
                'pred': pred,
                'true': true})
        pred_label.append(pred)
        true_label.append(true)
    return true_label, pred_label
```
